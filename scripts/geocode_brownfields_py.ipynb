```python
import pandas as pd
from geopy.geocoders import ArcGIS
import time

# 1. Load your dataset
df = pd.read_excel("Essex_County.xlsx")

# 2. Clean the messy city names
df['Clean_City'] = df['Site Municipality'].str.replace(' City', '').str.replace(' Twp', '').str.replace(' Village', '')

# 3. Switch to the ArcGIS geocoder (much smarter, still free)
geolocator = ArcGIS()

def get_lat_lon(row):
    address = f"{row['Line1 Address']}, {row['Clean_City']}, NJ"
    try:
        location = geolocator.geocode(address, timeout=10)
        if location:
            print(f" Found: {address}")
            return pd.Series([location.latitude, location.longitude])
    except:
        pass
    print(f" Missed: {address}")
    return pd.Series([None, None])

print("Finding coordinates using ArcGIS...")
df[['Latitude', 'Longitude']] = df.apply(get_lat_lon, axis=1)

# 4. Save the new file
df = df.drop(columns=['Clean_City'])
df.to_csv("Essex_County_with_Coordinates_V3.csv", index=False)
print("Done! File saved as V3.")
```
